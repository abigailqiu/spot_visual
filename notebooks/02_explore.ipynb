```python
import sys
sys.path.insert(0, "..")
import pandas as pd 
from src.config import SESSIONS_CSV
from gensim.models import Word2Vec

streams = pd.read_csv(SESSIONS_CSV)
sesh_list = streams.groupby("session_num")["spotify_track_uri"].agg(list).tolist() #first turn every bucket into list then turn whole series into list
```

```python
model = Word2Vec(sentences = sesh_list, vector_size = 32, window = 5, min_count = 2, sg = 1, seed = 42, epochs = 30, workers = 1)
```

```python
vector = model.wv['spotify:track:66tlRqExpkn4kRyuxelzs5']
print(vector)

print(len(model.wv), streams['spotify_track_uri'].nunique()) 

streams["label"] = streams["track_name"] + " - " + streams["artist_name"]
tracks = streams.drop_duplicates("spotify_track_uri").set_index("spotify_track_uri")["label"]
# print(tracks.head())
```

```python
track = 'spotify:track:66tlRqExpkn4kRyuxelzs5'
print(tracks[track])
similar_words = model.wv.most_similar(track, topn=10)
for uri, score in similar_words:
    print(tracks[uri], score)
```
